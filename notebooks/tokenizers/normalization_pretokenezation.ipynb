# Prepocessing for tokenization
Each tokenizer applies preopocessing to text:
- Normalization
- Pre-tokenization
- Model
- Postprocessor
![Preprocessing for tokenization](preprocessing.png)

## Normalization
General cleanup of the text, such as:
 - Removing needless whitespace
 - Lowercasing
 - Removing accents

```python
from transformers import AutoTokenizer
```

```python
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
print(type(tokenizer.backend_tokenizer))
```

```python
# Applying normalization to the text
tokenizer.backend_tokenizer.normalizer.normalize_str("Héllò hôw are ü?")
```

## Pre-tokenization
Split sentences into single words, punctuation, and special characters
It also provides offsets of each token.

```python
tokenizer.backend_tokenizer.pre_tokenizer.pre_tokenize_str("Hello, how are  you?")
```

We can apply different pre-tokenization of different tokenizers and see the difference.

```python
tokenizer = AutoTokenizer.from_pretrained("gpt2")
tokenizer.backend_tokenizer.pre_tokenizer.pre_tokenize_str("Hello, how are  you?")
```

```python
tokenizer = AutoTokenizer.from_pretrained("t5-small")
tokenizer.backend_tokenizer.pre_tokenizer.pre_tokenize_str("Hello, how are  you?")
```
