Creating corpus

```python
from datasets import load_dataset

dataset = load_dataset("wikitext", name="wikitext-2-raw-v1", split="train")


def get_training_corpus():
    for i in range(0, len(dataset), 1000):
        yield dataset[i : i + 1000]["text"]
```

```python
with open("wikitext-2.txt", "w", encoding="utf-8") as f:
    for i in range(len(dataset)):
        f.write(dataset[i]["text"] + "\n")
```

Building a WordPiece tokenizer from scratch

```python
from tokenizers import (
    decoders,
    models,
    normalizers,
    pre_tokenizers,
    processors,
    trainers,
    Tokenizer,
)

tokenizer = Tokenizer(models.WordPiece(unk_token="[UNK]"))
```

```python
tokenizer
```

## Normlization

```python
# Set normalizer as BestNormalizer for text normalization before tokenization. It lowercases the text and removes accents.
tokenizer.normalizer = normalizers.BertNormalizer(lowercase=True)
```

```python
print(tokenizer.normalizer.normalize_str("Héllò hôw are ü?"))
```

```python
# Or more from scratch approach 
tokenizer.normalizer = normalizers.Sequence(
    [normalizers.NFD(), normalizers.Lowercase(), normalizers.StripAccents()]
)
```

```python
print(tokenizer.normalizer.normalize_str("Héllò hôw are ü?"))
```

## Pre-tokenization 

```python
tokenizer.pre_tokenizer = pre_tokenizers.BertPreTokenizer()
```

or from scratch approach

```python
tokenizer.pre_tokenizer = pre_tokenizers.Whitespace()
```

```python
tokenizer.pre_tokenizer.pre_tokenize_str("Let's test my pre-tokenizer.")
```

```python
# And with white space split
pre_tokenizer = pre_tokenizers.WhitespaceSplit()
pre_tokenizer.pre_tokenize_str("Let's test my pre-tokenizer.")
```

Or with several tokenizers


```python
pre_tokenizer = pre_tokenizers.Sequence(
    [pre_tokenizers.WhitespaceSplit(), pre_tokenizers.Punctuation()]
)
pre_tokenizer.pre_tokenize_str("Let's test my pre-tokenizer.")
```
